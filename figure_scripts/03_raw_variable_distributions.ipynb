```python

import os
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from matplotlib.font_manager import FontProperties, fontManager

# Figure 3: Distribution of raw meteorological variables

# Repository root
ROOT_DIR = Path.cwd()

if ROOT_DIR.name == "figure_scripts":
    ROOT_DIR = ROOT_DIR.parent

# Input data
DATA_FILE = (
    ROOT_DIR
    / "dataset"
    / "data"
    / "processed"
    / "agroclimate_with_indicators.parquet"
)

# Output directory
OUTPUT_DIR = ROOT_DIR / "figure_scripts" / "outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

PNG_FILE = OUTPUT_DIR / "figure3_raw_variable_distributions.png"
PDF_FILE = OUTPUT_DIR / "figure3_raw_variable_distributions.pdf"

# Configure font
FONT_FILE = ROOT_DIR / "calibri.ttf"

if FONT_FILE.exists():
    fontManager.addfont(str(FONT_FILE))
    calibri = FontProperties(fname=str(FONT_FILE))
    plt.rcParams["font.family"] = calibri.get_name()
else:
    calibri = FontProperties(family="DejaVu Sans")
    plt.rcParams["font.family"] = "DejaVu Sans"

plt.rcParams["font.size"] = 13

# Load required columns
data = pd.read_parquet(
    DATA_FILE,
    columns=[
        "temperature_2m_mean",
        "temperature_2m_max",
        "temperature_2m_min",
        "precipitation_sum",
        "relative_humidity_2m_mean",
        "et0_fao_evapotranspiration"
    ]
)

# Configure variables, axis labels, and panel titles
variables = [
    ("temperature_2m_mean", "Temperature (°C)", "Mean Temperature"),
    ("temperature_2m_max", "Temperature (°C)", "Maximum Temperature"),
    ("temperature_2m_min", "Temperature (°C)", "Minimum Temperature"),
    ("precipitation_sum", "Precipitation (mm/day)", "Daily Precipitation"),
    ("relative_humidity_2m_mean", "Relative humidity (%)", "Mean Relative Humidity"),
    ("et0_fao_evapotranspiration", "ET₀ (mm/day)", "Reference Evapotranspiration")
]

# Create distribution plots
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(12, 12))
axes = axes.flatten()

for ax, (column, xlabel, panel_title) in zip(axes, variables):
    values = data[column].dropna()

    ax.hist(
        values,
        bins=50,
        edgecolor="black",
        linewidth=0.4
    )

    ax.set_title(
        panel_title,
        loc="left",
        fontproperties=calibri,
        fontsize=11,
        pad=8
    )

    ax.set_xlabel(xlabel, fontproperties=calibri, fontsize=13)
    ax.set_ylabel("Frequency", fontproperties=calibri, fontsize=13)
    ax.tick_params(axis="both", labelsize=11)

    for label in ax.get_xticklabels() + ax.get_yticklabels():
        label.set_fontproperties(calibri)
        label.set_fontsize(11)

    ax.grid(axis="y", linestyle="--", linewidth=0.4, alpha=0.4)

# Adjust layout and save figures
plt.tight_layout(h_pad=1.5, w_pad=1.2)

fig.savefig(PNG_FILE, dpi=300, bbox_inches="tight")
fig.savefig(PDF_FILE, format="pdf", bbox_inches="tight")

plt.show()
plt.close(fig)

print("Figure 3 saved:")
print(PNG_FILE)
print(PDF_FILE)
```
